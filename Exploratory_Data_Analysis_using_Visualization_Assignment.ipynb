```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Titanic.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.drop(columns=['Cabin'], inplace=True)
```

```python
df.head()
```

```python
#Q1. Create a countplot to show the number of passengers by gender.
```

```python
#Answer -
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Sex', palette='viridis')
plt.title('Number of Passengers by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()

#Interpretation: This chart shows the gender distribution on the ship. There were significantly more male passengers than female passengers
```

```python
#Q2. Create a bar chart to show the number of passengers in each passenger class using different colors.
```

```python
#Answer -
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Pclass', hue='Pclass', palette='Set2', legend=False)
plt.title('Number of Passengers in Each Passenger Class')
plt.xlabel('Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
plt.ylabel('Count')
plt.show()
#Interpretation: The majority of passengers were in the 3rd class, which was the most affordable option
```

```python
#Q3. Create a histogram to show the distribution of passenger age.
```

```python
#Answer -
plt.figure(figsize=(6, 5))
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Passenger Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
#Interpretation: Most passengers were young adults between the ages of 20 and 40
```

```python
#Q4. Create a boxplot to detect outliers in the fare column.
```

```python
#Answer -
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['Fare'], color='lightcoral')
plt.title('Boxplot of Passenger Fare')
plt.xlabel('Fare')
plt.show()
#Interpretation: There are several outliers in the fare column, with a few passengers paying extremely high prices (over 500) compared to the average
```

```python
#Q5. Create a scatterplot to
#a. show the relationship between age and fare.
#b. analyze the relationship between family size (SibSp + Parch) and fare.
```

```python
#Answer
#a-
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Fare', alpha=0.5)
plt.title('Relationship between Age and Fare')
plt.show()
#b-
df['FamilySize'] = df['SibSp'] + df['Parch']
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='FamilySize', y='Fare', color='green')
plt.title('Relationship between Family Size and Fare')
plt.show()

#Interpretation: There is no strong linear correlation between age and fare, but very high fares were mostly paid by adults. Larger families generally paid moderate fares
```

```python
#Q6. Create a countplot to show survival count by gender.
```

```python
#Answer -
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Survived', hue='Sex', palette='pastel')
plt.title('Survival Count by Gender')
plt.xticks([0, 1], ['Died', 'Survived'])
plt.show()
#Interpretation: A much higher percentage of females survived compared to males, likely due to the "women and children first" policy
```

```python
#Q7. Create a bar chart to show survival by passenger class.
```

```python
#Answer -
plt.figure(figsize=(5, 3))
sns.barplot(data=df, x='Pclass', y='Survived', palette='magma')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Probability')
plt.show()
#Interpretation: Passengers in 1st class had the highest probability of survival, while 3rd-class passengers had the lowest
```

```python
#Q8. Create a heatmap to show correlation between numerical variables.
```

```python
#Answer -
plt.figure(figsize=(10, 6))
# Selecting only numerical columns for correlation
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
#Interpretation: There is a notable negative correlation between Class (Pclass) and Fare, meaning as the class number increases (lower class), the fare decreases.
```

```python
#Q9. Identify which passenger class paid the highest fare on average.
```

```python
#Answer -
avg_fare = df.groupby('Pclass')['Fare'].mean()
print(f"Average Fare by Class:\n{avg_fare}")
```
