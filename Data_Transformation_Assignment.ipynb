```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, MinMaxScaler, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Load the provided data [cite: 13, 15]
data = pd.read_csv('/content/employee_productivity_dataset.csv')

# Note: The provided dataset has many duplicate rows (E001-E010 repeat multiple times).
# We should remove them first for a clean analysis.
df = data.drop_duplicates().reset_index(drop=True)
print("Initial Dataset (Duplicates Removed):")
display(df.head())
```

```python
#Q1. Handle Missing Values (Basic Cleaning)
#Fill missing values using the following:
#Age → Median
#Salary → Mean
#Hours_Worked_Per_Week → Median
#Performance_Score → Mean
#Display the dataset after handling missing values.
```

```python
#Answer - # Filling missing values [cite: 22, 23, 24, 25]
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Hours_Worked_Per_Week'] = df['Hours_Worked_Per_Week'].fillna(df['Hours_Worked_Per_Week'].median())
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].mean())

print("Dataset after handling missing values:")
display(df)
```

```python
#Q2. Label Encoding
#Convert these columns into numeric values using Label Encoding:
#Gender
#Department
#Show the updated columns.
```

```python
#Answer-
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
df['Department'] = le.fit_transform(df['Department'])

print("Updated columns after Label Encoding (Gender & Department):")
display(df[['Gender', 'Department']].head())
```

```python
#Q3. One-Hot Encoding
#Apply One-Hot Encoding on:
#Work_Mode
#Location
#Display the dataset and check how many new columns are created.
```

```python
#Answer -
# Applying One-Hot Encoding [cite: 36]
df_encoded = pd.get_dummies(df, columns=['Work_Mode', 'Location'])

print(f"Dataset after One-Hot Encoding. New columns created: {len(df_encoded.columns) - len(df.columns)}")
display(df_encoded.head())
```

```python
#Q4. Normalization (Min-Max Scaling)
#Normalize the following columns:
#Salary
#Hours_Worked_Per_Week
#Use MinMaxScaler and display results
```

```python
#Answer -
scaler_minmax = MinMaxScaler()
df_encoded[['Salary', 'Hours_Worked_Per_Week']] = scaler_minmax.fit_transform(df_encoded[['Salary', 'Hours_Worked_Per_Week']])

print("Columns after Min-Max Scaling:")
display(df_encoded[['Salary', 'Hours_Worked_Per_Week']].head())
```

```python
#Q5. Standardization (Scaling)
#Apply StandardScaler on:
#Age
#Pwskills
#Assignment Questions
#Projects_Completed
#Display the transformed values
```

```python
#Answer -
scaler_std = StandardScaler()
df_encoded[['Age', 'Projects_Completed']] = scaler_std.fit_transform(df_encoded[['Age', 'Projects_Completed']])

print("Transformed values after Standardization:")
display(df_encoded[['Age', 'Projects_Completed']].head())
```

```python
#Q6. Compare Scaling Methods
#Apply both:
#MinMaxScaler
#StandardScaler
#On the Salary column.
#Show both results side by side.
```

```python
#Answer - # Getting original salary (re-calculated mean for the comparison)
original_salary = df['Salary'].values.reshape(-1, 1)

compare_df = pd.DataFrame({
    'Original_Salary': original_salary.flatten(),
    'MinMax_Scaled': MinMaxScaler().fit_transform(original_salary).flatten(),
    'Standard_Scaled': StandardScaler().fit_transform(original_salary).flatten()
})

print("Comparison of Scaling Methods:")
display(compare_df.head())
```

```python
#Q7. Build Preprocessing Pipeline
#Create a pipeline that:
#Applies encoding to categorical columns
#Applies scaling to numerical columns
#Use:
#ColumnTransformer
#Pipeline

#Q8. Apply Pipeline
#Apply the pipeline on the dataset.
#Display:
#Transformed dataset
#Shape of final dataset
```

```python
# Define features
numeric_features = ['Age', 'Hours_Worked_Per_Week', 'Projects_Completed', 'Salary']
categorical_features = ['Gender', 'Department', 'Work_Mode', 'Location']

# Create transformers
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combine into a ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# Apply Pipeline
transformed_data = preprocessor.fit_transform(df)

print("Shape of final dataset:", transformed_data.shape)
print("Transformed Dataset (First 5 rows):")
print(transformed_data[:5])
```

```python
#Q9. Conceptual Question
#Why is scaling important in Python?

#Answer - Scaling ensures that features with larger numeric ranges (like Salary) do not dominate features with smaller ranges (like Age) during model training. Most algorithms (like KNN, SVM, and Neural Networks) use distance calculations; without scaling, the model would be biased toward the large-scale variables.
```

```python
#Q10. Conceptual Question
#Why do we convert categorical data into numerical form?

#Answer - Machine Learning models are mathematical equations that only understand numbers. They cannot perform calculations on text strings like "Remote" or "Finance." Converting them to numeric form (Encoding) allows the model to find patterns and correlations within the data
```
