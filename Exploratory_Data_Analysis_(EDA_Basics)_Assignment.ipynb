```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/House Price India.csv')
df.head()
```

```python
df.info()
```

```python
df.info()
#Interpretation:
#info() shows data types and missing value
```

```python
df.describe()
#Interpretation:
#describe() gives statistical summary (mean, std, min, max)
```

```python
df.isnull().sum()
```

```python
# If missing values exist
df = df.dropna()   # or use fillna()

#Interpretation:
#If null values are small → drop
#If large → use imputation
```

```python
#Q1. What is the average, median, and standard deviation of house prices?
```

```python
#Answer -
avg_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()

avg_price, median_price, std_price
#Interpretation: These metrics describe the central tendency and spread of property values. A large standard deviation indicates high price variability in the market.
```

```python
#Q2. Which number of bedrooms is most common?
```

```python
#Answer-
common_bedrooms = df['number of bedrooms'].mode()[0]
print(f"The most common number of bedrooms is: {common_bedrooms}")

sns.countplot(x='number of bedrooms', data=df)
plt.title('Distribution of Bedrooms')
plt.show()
#Interpretation: Identifying the most common room configuration helps understand the standard family size or property type most prevalent in the dataset.
```

```python
#Q3. Check if the price data is skewed.
```

```python
#Answer-
df['Price'].skew()
```

```python
sns.histplot(df['Price'], kde=True)
plt.show()

#Interpretation:
#Skew > 0 → Right skewed (common in real estate)
#Means few very expensive houses exist
```

```python
#Q4. What is the average price for each number of bedrooms?
```

```python
#Answer -
df.groupby('number of bedrooms')['Price'].mean().sort_values()
#Interpretation:
#Prices increase with bedrooms, but anomalies may exist (e.g., 10-bedroom houses not always most expensive).
```

```python
#Q5.  What is the relationship between living area and price? (Use : Area without Basement).
```

```python
#Answer -
# The column name from the CSV is 'Area of the house(excluding basement)'
sns.scatterplot(x='Area of the house(excluding basement)', y='Price', data=df)
plt.title('Living Area (Excl. Basement) vs Price')
plt.show()
#Interpretation:
#Strong positive correlation: larger living area → higher price. But diminishing returns after a threshold.
```

```python
#Q6. Identify any anomalies where houses have high prices but low areas.
```

```python
#Answer -
anomalies = df[(df['Price'] > df['Price'].quantile(0.90)) &
               (df['living area'] < df['living area'].quantile(0.25))]

anomalies
# Interpretation:
#These houses are overpriced or premium (location/luxury factors
```

```python
#Q7. Compare average price based on number of floors and houses with or without waterfront.
```

```python
#Answer -
df.groupby(['number of floors', 'waterfront present'])['Price'].mean()
#Interpretation:
#Waterfront houses usually cost more
#More floors may increase pric
```

```python
#Q8. Identify the minimum and maximum house price. What does this indicate?
```

```python
#Answer -
df['Price'].min(), df['Price'].max()

#Interpretation:
#Min price → affordable housing
#Max price → luxury segment
#Shows price range in market
```

```python
#Q9. Which location (zipcode/area) has the highest average price?
```

```python
#Answer -
df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False)
#Interpretation:
#Top Postal Code = premium location
#Useful for investment decisions
```
